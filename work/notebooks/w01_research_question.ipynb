## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
import pandas as pd

# Load dataset
df = pd.read_csv('https://raw.githubusercontent.com/PaNavar369/Internship/main/data/raw/content_refresh_anonymized.csv')

# Number 1: Age gap between stable/growing pages vs dropping pages
avg_age_down = df[df['trend_direction'] == 'down']['content_age_days'].mean()
avg_age_up = df[df['trend_direction'] == 'up']['content_age_days'].mean()
print(f"1. Age Gap: Pages trending 'up' average {avg_age_up:.0f} days old vs 'down' pages at {avg_age_down:.0f} days old.")


ghost_pages = df[(df['impressions_90d'] >= 1000) & (df['clicks_90d'] == 0)]
pct_ghost = len(ghost_pages) / len(df)
print(f"2. Zero-Click High-Impression Pages: {len(ghost_pages):,} pages ({pct_ghost:.1%}) received over 1,000 impressions in 90 days but ZERO clicks.")

# Number 2: Word count similarity between up and down pages (dispelling a myth)
med_wc_down = df[df['trend_direction'] == 'down']['word_count'].median()
med_wc_up = df[df['trend_direction'] == 'up']['word_count'].median()
print(f"2. Word Count Lever: Median word count for 'up' ({med_wc_up:.0f}) vs 'down' ({med_wc_down:.0f}) is almost identical.")

# Number 3: Non-linear CTR drop (Position 1 vs Position 10)
visible = df[df['impressions_90d'] >= 100]
pos1_ctr = visible[visible['avg_position'] <= 1.5]['ctr'].mean()
pos10_ctr = visible[(visible['avg_position'] >= 9.5) & (visible['avg_position'] <= 10.5)]['ctr'].mean()
print(f"3. Non-linear CTR: Average CTR drops from {pos1_ctr:.1%} at Position 1 down to {pos10_ctr:.1%} at Position 10.")
```
