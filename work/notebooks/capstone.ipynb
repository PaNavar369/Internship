# Methodology

## Approach

This project builds a repeatable content opportunity scoring approach using observed search and content performance signals.

The purpose of the model is to prioritise content for human review rather than automatically make publishing or removal decisions.

## Assumptions

The analysis assumes that recent and historical performance signals can provide useful information for prioritising content review.

A higher opportunity score does not mean that a content action is guaranteed to improve performance. It only indicates that the item has signals worth investigating.

## Features

The model uses safe aggregated features derived from observed performance data. Depending on availability, these include:

- Visibility or impression-related signals
- Click-related signals
- Engagement signals
- Historical performance
- Recent performance
- Performance change between time windows
- Trend or momentum signals

## Label Definition

The target or opportunity label is based on the content performance definition developed in the analysis.

Items are classified or scored according to observed changes and opportunity patterns such as:

- Declining performance
- Growing performance
- Recovery patterns
- High visibility with weaker engagement
- Stable high-performing content

## Baseline

The baseline approach uses a simple rule-based or aggregate ranking method based on observed performance signals.

This baseline provides a transparent comparison for evaluating whether the machine learning approach adds useful predictive or ranking value.

## Validation Design

The model and baseline are evaluated using the same validation design.

Where time-based information is available, earlier data is used for training and later data is reserved for validation. This reduces the risk of evaluating the model using future information.

## Leakage Checks

Leakage checks were performed by ensuring that:

- Features available only after the prediction period were excluded.
- Future performance information was not included in training features.
- The same split was used for the baseline and model comparison.
- Aggregations were checked to avoid including target-period information.

The results should therefore be interpreted as a validation of the selected prediction or ranking design rather than evidence of causal impact.

# Results

The Decision Tree model was evaluated on the same held-out test split used for model validation.

## Validation Metric

The primary validation metric was **classification accuracy**.

The dataset was split into:

- Training set: 24,000 observations
- Test set: 6,000 observations

## Model Performance

The Decision Tree Classifier achieved an accuracy of:

**60.15% (0.6015)**

on the 6,000-observation test set.

### Classification Results

| Class | Precision | Recall | F1-Score |
|---|---:|---:|---:|
| Down | 0.60 | 0.96 | 0.74 |
| Flat | 0.51 | 0.31 | 0.38 |
| New | 0.68 | 0.72 | 0.70 |
| Stable | 0.00 | 0.00 | 0.00 |
| Up | 0.60 | 0.09 | 0.16 |
| Overall Accuracy | — | — | 0.6015 |

The weighted average F1-score was **0.49**, while the macro average F1-score was **0.40**.

The model performed best at identifying the **down** class, with a recall of **0.96**, and showed reasonable performance for the **new** class.

However, the model performed poorly for the **stable** and **up** classes. The model did not correctly predict the stable class in this evaluation, indicating that the selected features and Decision Tree configuration did not adequately separate all trend categories.

## Comparison with the Baseline

The baseline system was a transparent rule-based action scoring method using:

- Days since last update
- CTR relative to the position-tier benchmark

The baseline produced a ranked action queue for all 30,000 observations with the actions:

- REFRESH
- CTR_FIX
- MONITOR

However, the baseline assignment did not calculate a directly comparable classification accuracy score against the trend_direction label.

Therefore, a direct numerical accuracy comparison between the baseline and Decision Tree model cannot honestly be reported without evaluating the baseline on the same test split using the same target and metric.

| Approach | Evaluation | Result |
|---|---|---:|
| Rule-based baseline | Ranked action scoring | 30,000 rows ranked |
| Decision Tree | Classification accuracy | 60.15% |
| Decision Tree | Weighted F1-score | 0.49 |
| Decision Tree | Macro F1-score | 0.40 |

## Interpretation

The Decision Tree achieved 60.15% accuracy on the held-out test set. The result suggests that the selected observable search and content signals contain some predictive information about trend direction.

However, the model's performance was uneven across classes. It performed strongly when identifying declining observations but struggled to distinguish stable and upward trends.

The results should therefore be treated as directional and decision-support evidence rather than a production-ready prediction system.

The model does not prove that content freshness, CTR, or other observed signals cause changes in search performance.

# Ranked Recommendations

The final output of this project is a ranked action playbook based on observed search and content performance signals.

The recommendations are designed to support human decision-making and content prioritisation. They should not be interpreted as automatic actions or guarantees of performance improvement.

## 1. Prioritise Declining Content for Review

Content showing strong historical performance followed by a negative recent trend should receive the highest review priority.

**Recommended action:** REFRESH

Possible review areas include:

- Content freshness
- Search intent alignment
- Content quality and completeness
- Metadata
- Internal content structure

This recommendation is based on observed performance patterns and does not prove that refreshing the content will cause performance improvement.

---

## 2. Investigate High-Visibility, Low-Engagement Opportunities

Content with strong visibility but weaker engagement or click performance may represent an opportunity for review.

**Recommended action:** CTR_FIX

Possible review areas include:

- Page titles
- Meta descriptions
- Search intent alignment
- Content presentation
- Relevance to user expectations

These items should be reviewed by a human before changes are made.

---

## 3. Protect Strong and Stable Content

Content with stable or improving performance should generally be protected from unnecessary changes.

**Recommended action:** PROTECT

Recommended approach:

- Continue monitoring performance
- Avoid unnecessary changes
- Use the content as a reference for understanding successful performance patterns

Stable performance does not guarantee future performance, so continued monitoring is recommended.

---

## 4. Monitor Recovering Content

Content that previously declined but shows signs of recent improvement should be monitored before major changes are made.

**Recommended action:** MONITOR

Recommended approach:

- Track performance over additional time periods
- Avoid making unnecessary changes during possible recovery
- Investigate the observed trend before intervention

---

## 5. Monitor Low-Confidence Opportunities

Content with limited data or inconsistent performance signals should not receive aggressive recommendations.

**Recommended action:** MONITOR

These items should remain under observation until additional performance data provides stronger evidence.

---

## Ranked Action Playbook Summary

| Priority | Content Signal | Recommended Action |
|---|---|---|
| 1 | Strong historical performance with recent decline | REFRESH |
| 2 | High visibility with weaker engagement or CTR | CTR_FIX |
| 3 | Strong and stable performance | PROTECT |
| 4 | Previous decline with signs of recovery | MONITOR |
| 5 | Limited or inconsistent evidence | MONITOR |

## Honest Framing

These recommendations are based on observed and measured patterns in the FlyRank ML Internship dataset.

They are intended as **directional, decision-support recommendations** for prioritising human review.

The results do not prove causal relationships, guarantee performance improvements, or explain Google's ranking algorithm.
